### Table Consolidation - copy for a simple script for product development

```python
import pandas as pd
```

```python
stop_list = pd.read_csv("stops.txt")
stop_list.head(20)
```

```python
timing_list = pd.read_csv("stop_times.txt")
timing_list.head(20)
```

```python
timing_consol = pd.merge(stop_list, timing_list, on = "stop_id")
timing_consol.head(20)
```

```python
print(f"Consolidated length : {len(timing_consol)}\tNumber of stops : {len(stop_list)}\tStop times table : {len(timing_list)}")
```

```python
trip_list = pd.read_csv("trips.txt")
trip_list.head(20)
```

```python
timing_consol = pd.merge(timing_consol, trip_list, on = "trip_id")
timing_consol.head(20)
```

```python
routes_list = pd.read_csv("routes.txt")
routes_list.head(20)
```

```python
timing_consol = pd.merge(timing_consol, routes_list, on = "route_id")
timing_consol.head(20)
```

```python
timing_consol.columns
```

```python
drop_col = ['zone', 'trip_id', 'stop_sequence', 'timepoint', 'route_id', 'service_id', 'shape_id', 'direction_id', 'agency_id', 'route_type', 'route_color', 'route_text_color', 'fare_rule']
timing_consol = timing_consol.drop(columns = drop_col, axis = 1)
timing_consol.head(20)
```

### Map marking prototype

```python
import folium
```

```python
centre_lat = stop_list["stop_lat"].mean()
centre_lon = stop_list["stop_lon"].mean()

m = folium.Map(location=[centre_lat, centre_lon], zoom_start=13)

for i, row in stop_list.iterrows() :
    folium.Marker(
        location=[row['stop_lat'], row['stop_lon']],
        popup=row['stop_name'], # The text that appears when clicked
        icon=folium.Icon(color='red', icon='info-sign')
    ).add_to(m)

m
```

### Import to CSV the app will read from this csv file

```python
timing_consol.to_csv('consolidated_timing.csv', index=False)
```
